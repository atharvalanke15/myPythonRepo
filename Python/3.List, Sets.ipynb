## Set
### Unordered
### No indexing
### No duplicates
### No mutables

```python
myset=set()
print(type(myset))
print(myset)
print(len(myset))
```

```python
myset={1, 2, 3, 4, 5,5, 'a', 'b', 'c', 'd', 'e'}
print(myset)
print(len(myset))
```

```python
myset=set()
myset={1,1,2,3,4,4,4,5,6}
print(myset)
myset.add(7)
print(myset)
```

```python
myset1={1, 2, 3, 4, 5}
myset2={4, 5, 6, 7, 8, "ATHARVA"}
myset2.difference(myset1)
myset2.intersection_update(myset1)
print(myset2)
myset2.intersection(myset1)
```

## Tuples - Built in data type used to store a collection of items
### Ordered- items keep their position
### Immutable - you cannot change values after creation
### Allows duplicates

```python
mytuple=tuple()
print(type(mytuple))
```

```python
mytuple=(1, 2, 3, 4, 5)
print(mytuple)
```

```python
mytuple=(True, False, True, False, True, False, True, False, True, False)
print(mytuple)
```

```python
mytuple=("atharva" ,1996, "Pune", 1996, "India", 1947)
print(mytuple)
print(mytuple.count(1996))
print(mytuple.index("Pune"))
print(mytuple[0])
print(mytuple[1])
print(mytuple[5])
print(mytuple.index("India"))
print(mytuple.index(1947))
```
